```python
# ==== Compare WITH anomalies vs AFTER DROPPING anomalies (same output structure) ====
import pandas as pd, numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    r2_score, mean_absolute_error,
    precision_score, recall_score, f1_score, average_precision_score
)
from sklearn.base import clone
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import (
    IsolationForest, RandomForestRegressor, ExtraTreesRegressor, GradientBoostingRegressor,
    VotingRegressor, StackingRegressor
)
from sklearn.linear_model import LinearRegression, Ridge, BayesianRidge, HuberRegressor
from sklearn.tree import DecisionTreeRegressor
from xgboost import XGBRegressor

# ---------- 0) Load + minimal features ----------
PATH = r"C:\Users\israt\HD lab\My Project\Dataset\Airsage Dataset\NebraskaLakes_Cleaned_round.csv"
df = pd.read_csv(PATH, low_memory=False)

df["date"] = pd.to_datetime(df["date"], format="%Y%m%d", errors="coerce")
df["count"] = pd.to_numeric(df["count"], errors="coerce")      # target
df["duration"] = pd.to_numeric(df["duration"], errors="coerce")
df = df.dropna(subset=["date","count","duration"]).copy()

# simple time features
df["month"] = df["date"].dt.month
df["dow"] = df["date"].dt.dayofweek
df["is_weekend"] = (df["dow"] >= 5).astype(int)

# --- meta to print later (NO effect on modeling/splits) ---
meta_all = df[["poi_id", "poi", "date"]].copy()
meta_all["date_str"] = meta_all["date"].dt.strftime("%Y%m%d")

feat_cols = [c for c in df.select_dtypes(include=[np.number]).columns
             if c not in {"count","anomaly_label","anomaly_score"}]
X_all = df[feat_cols].fillna(0).to_numpy(np.float32)
y_all = df["count"].to_numpy(np.float32)

# ---------- 1) Models to evaluate ----------
candidates = [
    ("rf",    RandomForestRegressor(n_estimators=600, max_depth=None,
                                    min_samples_split=2, min_samples_leaf=1,
                                    max_features="sqrt", n_jobs=-1, random_state=42)),
    ("xgb",   XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=6,
                           subsample=0.8, colsample_bytree=0.8, reg_lambda=1.0,
                           n_jobs=-1, random_state=42, tree_method="hist")),
    ("etr",   ExtraTreesRegressor(n_estimators=600, max_depth=None,
                                  min_samples_leaf=1, max_features="sqrt",
                                  n_jobs=-1, random_state=42)),
    ("gbr",   GradientBoostingRegressor(random_state=42, n_estimators=300, learning_rate=0.05, max_depth=3)),
    ("dt",    DecisionTreeRegressor(max_depth=12, random_state=42)),
    ("lr",    LinearRegression()),
    ("ridge", Ridge(alpha=1.0, random_state=42)),
    ("bayes", BayesianRidge()),
    ("huber", HuberRegressor(epsilon=1.35)),
]

# ---------- 2) Helper: rank models on one split ----------
def rank_models_on_split(X_tr, X_te, y_tr, y_te, candidates, label):
    TH = float(np.median(y_tr))  # threshold for P/R/F1/PR-AUC
    rows, fitted = [], {}
    for name, mdl in candidates:
        m = clone(mdl)
        m.fit(X_tr, y_tr)
        fitted[name] = m
        pred = m.predict(X_te)

        r2  = r2_score(y_te, pred)
        mae = mean_absolute_error(y_te, pred)

        y_true = (y_te > TH).astype(int)
        y_hat  = (pred  > TH).astype(int)

        P  = precision_score(y_true, y_hat, zero_division=0)
        R  = recall_score(y_true, y_hat, zero_division=0)
        F1 = f1_score(y_true, y_hat, zero_division=0)
        PR = average_precision_score(y_true, pred)

        rows.append({"Scenario": label, "Model": name.upper(),
                     "P": P, "R": R, "F1": F1, "PR_AUC": PR, "R2": r2, "MAE": mae})

    rank_df = (pd.DataFrame(rows)
                 .sort_values(["F1","PR_AUC","R2","MAE","P","R"],
                              ascending=[False,False,False,True,False,False])
                 .reset_index(drop=True))
    return rank_df

# ---------- 3) Scenario A: WITH anomalies (no drop) ----------
X_tr_A, X_te_A, y_tr_A, y_te_A = train_test_split(X_all, y_all, test_size=0.2, random_state=42)
rank_with = rank_models_on_split(X_tr_A, X_te_A, y_tr_A, y_te_A, candidates, label="WITH_ANOMALIES")

# ---------- 4) Scenario B: AFTER DROPPING anomalies (IsolationForest prefilter) ----------
PREFILTER_CONTAM = 0.10
Z = StandardScaler().fit_transform(X_all)  # unsupervised, no target leakage
gate = IsolationForest(contamination=PREFILTER_CONTAM, random_state=42).fit(Z)
keep_mask = (gate.predict(Z) == 1)  # 1=normal, -1=anomaly
X_clean, y_clean = X_all[keep_mask], y_all[keep_mask]

# align meta with cleaned rows (printing only; no modeling change)
meta_clean = meta_all[keep_mask].reset_index(drop=True)

print(f"Dropped {len(y_all) - len(y_clean)} anomalies "
      f"({100*(1 - len(y_clean)/len(y_all)):.1f}%) using IsolationForest prefilter.")

# Split Scenario B WITH meta (does not alter X/y permutation)
X_tr_B, X_te_B, y_tr_B, y_te_B, m_tr_B, m_te_B = train_test_split(
    X_clean, y_clean, meta_clean, test_size=0.2, random_state=42
)
rank_wo = rank_models_on_split(X_tr_B, X_te_B, y_tr_B, y_te_B, candidates, label="NO_ANOMALIES")

# ---------- 5) Print BOTH tables (identical structure) ----------
pd.set_option("display.max_rows", None)
pd.set_option("display.width", None)

print("\n=== RANKED (WITH anomalies) ===")
print(rank_with[["Model","P","R","F1","PR_AUC","R2","MAE"]]
      .to_string(index=False, float_format=lambda x: f"{x:.3f}"))

print("\n=== RANKED (AFTER DROPPING anomalies) ===")
print(rank_wo[["Model","P","R","F1","PR_AUC","R2","MAE"]]
      .to_string(index=False, float_format=lambda x: f"{x:.3f}"))

# ---------- 6) Optional: side-by-side comparison by Model ----------
cmp = (rank_with.merge(rank_wo, on="Model", suffixes=("_WITH","_NOANOM"))
               .loc[:, ["Model",
                        "F1_WITH","F1_NOANOM",
                        "PR_AUC_WITH","PR_AUC_NOANOM",
                        "R2_WITH","R2_NOANOM",
                        "MAE_WITH","MAE_NOANOM",
                        "P_WITH","P_NOANOM","R_WITH","R_NOANOM"]])

print("\n=== SIDE-BY-SIDE (same models) ===")
print(cmp.to_string(index=False, float_format=lambda x: f"{x:.3f}"))

# ---------- 7) Ensembles (Top-3) + Top-3 hybrid avg ----------
# Take Top-3 models from NO_ANOMALIES ranking
top3_noanom = rank_wo["Model"].head(3).tolist()   # e.g., ["XGB","RF","ETR"]
top3 = list(dict.fromkeys(top3_noanom))
assert len(top3) == 3, f"Expected 3 unique models in top3, got {top3}"

# We'll ensemble on the NO_ANOMALIES split
X_tr, X_te, y_tr, y_te = X_tr_B, X_te_B, y_tr_B, y_te_B

# Map model name -> estimator from `candidates` (uppercase keys)
cand_map = {name.upper(): est for name, est in candidates}
for nm in top3:
    assert nm in cand_map, f"{nm} not in candidates: {list(cand_map)}"

# Base learners for ensembles (must be uppercase names)
bases = [(nm, clone(cand_map[nm])) for nm in top3]

# Small cache of fitted base candidates (keys are original names: 'rf','xgb',...)
fitted = {name: clone(est).fit(X_tr, y_tr) for name, est in candidates}

# Voting (equal weights across Top-3)
vote = VotingRegressor(bases).fit(X_tr, y_tr)
pred_vote = vote.predict(X_te)

# Stacking with several meta-learners
stack_lin = StackingRegressor(
    estimators=bases,
    final_estimator=LinearRegression(),
    cv=3, passthrough=True
).fit(X_tr, y_tr)
pred_stack_lin = stack_lin.predict(X_te)

stack_ridge = StackingRegressor(
    estimators=bases,
    final_estimator=Ridge(alpha=1.0, random_state=42),
    cv=3, passthrough=True
).fit(X_tr, y_tr)
pred_stack_ridge = stack_ridge.predict(X_te)

stack_gbr = StackingRegressor(
    estimators=bases,
    final_estimator=GradientBoostingRegressor(random_state=42, n_estimators=200, learning_rate=0.05, max_depth=3),
    cv=3, passthrough=True
).fit(X_tr, y_tr)
pred_stack_gbr = stack_gbr.predict(X_te)

# ---- Hybrid (Top-3 equal-weight average) ----
def _predict_from_name(nm_upper: str):
    key_l = nm_upper.lower()
    if key_l in fitted:
        return fitted[key_l].predict(X_te)
    return clone(cand_map[nm_upper]).fit(X_tr, y_tr).predict(X_te)

_top3_preds = [_predict_from_name(nm) for nm in top3]     # e.g., preds from ["XGB","RF","ETR"]
pred_avg = np.mean(np.column_stack(_top3_preds), axis=1)

# ---------- 8) Report ----------
def show(name, yt, yp):
    print(f"{name:<24} R2={r2_score(yt, yp):.4f}  MAE={mean_absolute_error(yt, yp):.4f}")

print("\n=== Ensembles with Top-3 bases (NO_ANOMALIES split) ===")
show("Voting (Top-3, equal)", y_te, pred_vote)
show("Stack (Linear meta)",   y_te, pred_stack_lin)
show("Stack (Ridge meta)",    y_te, pred_stack_ridge)
show("Stack (GBR meta)",      y_te, pred_stack_gbr)
show("Hybrid (Top-3 avg)",    y_te, pred_avg)

```

```python
# ================== SIMPLE (tunable): stronger weekend/summer separation ==================
import numpy as np, pandas as pd
from sklearn.base import clone

def forecast_compare_simple(
    df_full, feat_cols, poi_id=None, poi_name=None,
    start_date=None, days=7,
    which="hybrid",                 # 'ridge'|'lin'|'gbr'|'voting'|'hybrid'|'all'
    round_int=False,
    season_boost=True,              # turn on/off seasonality multiplier for WITH_EXTRAS
    season_mode="product",          # 'product' (strong) or 'blend' (soft)
    w_month=0.75,                   # weight for month in 'blend' mode
    weekend_power=1.2,              # raise DOW factor to this power (>=1 increases weekend lift)
    month_power=1.1,                # raise MONTH factor to this power (>=1 increases summer lift)
    cap=(0.6, 1.6)                  # clamp season factor to [min,max] to avoid extremes
):
    # ---- tiny helpers ----
    def _poi(df, pid, pname):
        if pid is not None:
            s = df[df["poi_id"]==int(pid)]
            return int(pid), (s["poi"].iloc[0] if not s.empty and "poi" in s else str(pid))
        key = str(pname).strip().upper()
        s = df[df["poi"].astype(str).str.upper()==key]
        if s.empty: s = df[df["poi"].astype(str).str.upper().str.startswith(key)]
        if s.empty: raise ValueError(f"POI '{pname}' not found.")
        return int(s["poi_id"].iloc[0]), str(s["poi"].iloc[0])

    def _last(df, pid, asof=None):
        d = df[df["poi_id"]==pid].sort_values("date")
        if d.empty: raise ValueError(f"No history for poi_id={pid}.")
        if asof is not None:
            d2 = d[d["date"]<=asof]
            if not d2.empty: d = d2
        return d.iloc[-1], d

    # trained column layout
    probe = None
    for nm in ("stack_ridge","stack_lin","stack_gbr","vote"):
        if nm in globals():
            probe = globals()[nm]; break
    trained_cols = list(getattr(probe, "feature_names_in_", feat_cols))

    # predictors
    def _predict(Xf):
        preds = {}
        if which in ("ridge","all") and "stack_ridge" in globals(): preds["Stack_Ridge"]=float(stack_ridge.predict(Xf)[0])
        if which in ("lin","all")   and "stack_lin"   in globals(): preds["Stack_Lin"]  =float(stack_lin.predict(Xf)[0])
        if which in ("gbr","all")   and "stack_gbr"   in globals(): preds["Stack_GBR"]  =float(stack_gbr.predict(Xf)[0])
        if which in ("voting","all")and "vote"        in globals(): preds["Voting"]     =float(vote.predict(Xf)[0])
        if which in ("hybrid","all"):
            if "top3" not in globals(): raise RuntimeError("Need global 'top3' from section 7.")
            hy=[]
            for nm in top3:
                mdl = globals()["fitted"].get(nm.lower())
                if mdl is None: mdl = clone(globals()["cand_map"][nm]).fit(globals()["X_tr"], globals()["y_tr"])
                hy.append(float(mdl.predict(Xf)[0]))
            preds["Hybrid"]=float(np.mean(hy))
        if not preds: raise RuntimeError("No fitted model available.")
        return (float(np.mean(list(preds.values()))) if which=="all"
                else (preds.get("Hybrid") or preds.get("Voting") or preds.get("Stack_GBR") or preds.get("Stack_Ridge") or preds.get("Stack_Lin")))

    # feature row builders
    def _row_with(last_row, cur, last_count, recent):
        x = {c:0.0 for c in trained_cols}
        for c in trained_cols:
            if c in last_row.index:
                try: x[c]=float(last_row[c])
                except: pass
        if "poi_id" in x: x["poi_id"]=float(last_row["poi_id"])
        # calendar + Fourier
        if "month" in x: x["month"]=float(cur.month)
        if "dow" in x: x["dow"]=float(cur.dayofweek)
        if "is_weekend" in x: x["is_weekend"]=float(int(cur.dayofweek in (5,6)))
        if ("sin_y" in x) or ("cos_y" in x):
            doy=float(cur.timetuple().tm_yday)
            if "sin_y" in x: x["sin_y"]=float(np.sin(2*np.pi*doy/365.0))
            if "cos_y" in x: x["cos_y"]=float(np.cos(2*np.pi*doy/365.0))
        # lags
        if "lag_1" in x: x["lag_1"]=float(last_count)
        if "lag_7" in x: x["lag_7"]=float(recent[0] if len(recent)>=7 else last_count)
        if "roll7" in x and len(recent)>0: x["roll7"]=float(np.mean(recent))
        return pd.DataFrame([[x[c] for c in trained_cols]], columns=trained_cols)

    def _row_noextras(last_row):
        x = {c:0.0 for c in trained_cols}
        for c in trained_cols:
            if c in last_row.index:
                try: x[c]=float(last_row[c])
                except: pass
        if "poi_id" in x: x["poi_id"]=float(last_row["poi_id"])
        # intentionally DO NOT update calendar/Fourier/lags
        return pd.DataFrame([[x[c] for c in trained_cols]], columns=trained_cols)

    # season profiles (data-driven)
    def _season_profiles(hist):
        gmed = float(hist["count"].median()) or 1.0
        dow_rel = (hist.groupby(hist["date"].dt.dayofweek)["count"].median()/gmed).to_dict()
        mon_rel = (hist.groupby(hist["date"].dt.month)["count"].median()/gmed).to_dict()
        return dow_rel, mon_rel

    def _season_factor(dow_rel, mon_rel, dt):
        d = int(dt.dayofweek); m = int(dt.month)
        d_fac = float(dow_rel.get(d,1.0))
        m_fac = float(mon_rel.get(m,1.0))
        if season_mode == "blend":
            fac = (1.0 - w_month)*d_fac + w_month*m_fac
        else:  # 'product' (stronger)
            fac = (d_fac**weekend_power) * (m_fac**month_power)
        # clamp
        lo, hi = cap
        return max(lo, min(hi, fac))

    # -------- run both modes --------
    pid, pname = _poi(df_full, poi_id, poi_name)
    if start_date is None:
        last_d = df_full[df_full["poi_id"]==pid]["date"].max()
        if pd.isna(last_d): raise ValueError(f"No history for poi_id={pid}.")
        start = (last_d + pd.Timedelta(days=1)).normalize()
    else:
        start = pd.to_datetime(str(start_date), errors="coerce")
        if pd.isna(start): raise ValueError("Bad start_date; use YYYY-MM-DD or YYYYMMDD")

    last_row, hist = _last(df_full, pid, asof=start - pd.Timedelta(days=1))
    last_count0 = float(last_row["count"])
    before = hist[hist["date"] <= (start - pd.Timedelta(days=1))]
    recent0 = before["count"].tail(7).astype(float).tolist() if not before.empty else [last_count0]
    if not recent0: recent0=[last_count0]

    # season dicts
    dow_rel, mon_rel = _season_profiles(hist)

    outputs=[]
    for mode in ("WITH_EXTRAS","NO_EXTRAS"):
        rows=[]; cur=start; last_row_m=last_row.copy(); last_count=last_count0; recent=list(recent0)
        for _ in range(int(days)):
            Xf = _row_with(last_row_m, cur, last_count, recent) if mode=="WITH_EXTRAS" else _row_noextras(last_row_m)
            yhat = _predict(Xf)
            if mode=="WITH_EXTRAS" and season_boost:
                yhat *= _season_factor(dow_rel, mon_rel, cur)
            rows.append({
                "mode":mode,"poi_id":pid,"poi":pname,"date":cur,"date_str":cur.strftime("%Y-%m-%d"),
                "predicted": int(round(yhat)) if round_int else float(yhat)
            })
            if mode=="WITH_EXTRAS":
                recent.append(yhat); 
                if len(recent)>7: recent.pop(0)
            last_count=yhat
            last_row_m = last_row_m.copy()
            if "count" in last_row_m.index: last_row_m["count"]=last_count
            cur += pd.Timedelta(days=1)
        outputs.append(pd.DataFrame(rows))
    out = pd.concat(outputs, ignore_index=True)
    return out[["mode","poi_id","poi","date","date_str","predicted"]]
```

```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

def scatter_index_overlay(y_true, y_pred, title, x=None, s=8, alpha=0.6, save_path=None):
    """
    Overlaid scatter: Actual and Predicted vs index (or provided x).
    - y_true, y_pred: 1D arrays of equal length
    - x: optional x-axis values (default = 0..N-1)
    - s, alpha: marker size and transparency
    - save_path: optional path to save the figure
    """
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    assert y_true.shape == y_pred.shape, "y_true and y_pred must have same shape"

    if x is None:
        x = np.arange(len(y_true))
    else:
        x = np.asarray(x).ravel()
        assert len(x) == len(y_true), "x must match length of y arrays"

    plt.figure(figsize=(10, 4))
    plt.scatter(x, y_true, s=s, alpha=alpha, label="Actual")
    plt.scatter(x, y_pred, s=s, alpha=alpha, label="Predicted")
    plt.title(title)
    plt.xlabel("Index")
    plt.ylabel("Value")
    plt.legend()
    plt.tight_layout()

    if save_path:
        Path(save_path).parent.mkdir(parents=True, exist_ok=True)
        plt.savefig(save_path, dpi=150)
    plt.show()
    plt.close()
```

```python
# Fast variant (to finish under time):
# - Limits training window to the last 800 days for the selected POI
# - Keeps essential lags/rollings
# - Uses HistGradientBoosting with fewer iterations
import pandas as pd, numpy as np, warnings
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import r2_score, mean_absolute_error

CSV_PATH = r"C:\Users\israt\HD lab\My Project\Dataset\Airsage Dataset\NebraskaLakes_Cleaned_round.csv"
POI_ID = None
FORECAST_DAYS = 14
WINDOW_DAYS = 800

CAL_KEEP = ["month","day","dow","week","is_weekend","season",
            "sin_doy","cos_doy","sin_dow","cos_dow"]

def _season_from_month(m):
    if m in (12,1,2): return 1
    if m in (3,4,5):  return 2
    if m in (6,7,8):  return 3
    return 4

def _make_calendar(d):
    d = d.copy()
    d["month"] = d["date"].dt.month.astype(np.int16)
    d["day"] = d["date"].dt.day.astype(np.int16)
    d["dow"] = d["date"].dt.weekday.astype(np.int16)
    d["week"] = d["date"].dt.isocalendar().week.astype(np.int16)
    d["is_weekend"] = d["dow"].isin([5,6]).astype(np.int8)
    d["season"] = d["month"].apply(_season_from_month).astype(np.int8)
    doy = d["date"].dt.dayofyear.astype(int)
    d["sin_doy"] = np.sin(2*np.pi*doy/365).astype(np.float32)
    d["cos_doy"] = np.cos(2*np.pi*doy/365).astype(np.float32)
    d["sin_dow"] = np.sin(2*np.pi*(d["dow"]+1)/7).astype(np.float32)
    d["cos_dow"] = np.cos(2*np.pi*(d["dow"]+1)/7).astype(np.float32)
    return d

def _build_lags(g):
    g = g.sort_values("date").copy()
    for k in [1,7,14,28]:
        g[f"lag_{k}"] = g["count"].shift(k)
    g["roll_mean_7"]  = g["count"].rolling(7, min_periods=1).mean().shift(1)
    g["roll_mean_28"] = g["count"].rolling(28, min_periods=1).mean().shift(1)
    g["roll_std_7"]   = g["count"].rolling(7, min_periods=1).std(ddof=0).shift(1)
    g["roll_std_28"]  = g["count"].rolling(28, min_periods=1).std(ddof=0).shift(1)
    return g

def _time_split(h, test_size=0.2):
    n = len(h)
    n_te = max(1, int(round(n*test_size)))
    n_tr = n - n_te
    return h.iloc[:n_tr].copy(), h.iloc[n_tr:].copy()

def _top10_original_numeric(tr, target="count", exclude=()):
    num_cols = tr.select_dtypes(include=[np.number]).columns.tolist()
    ex = set(exclude) | {target, "poi_id"}
    base = [c for c in num_cols if c not in ex]
    if not base:
        return []
    corr = tr[base].corrwith(tr[target]).abs().sort_values(ascending=False)
    return corr.head(10).index.tolist()

with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    df = pd.read_csv(CSV_PATH, low_memory=False)

df["date"] = pd.to_datetime(df["date"].astype(str), format="%Y%m%d", errors="coerce")
df = df.dropna(subset=["date"])
df["count"] = pd.to_numeric(df["count"], errors="coerce")
df = df.dropna(subset=["count"])

if "poi_id" not in df.columns:
    if "poi" in df.columns:
        df["poi_id"] = pd.factorize(df["poi"].astype(str))[0].astype(int)
    else:
        df["poi_id"] = 0

df = df.sort_values(["poi_id","date"]).reset_index(drop=True)
df = _make_calendar(df)
df = df.groupby("poi_id", group_keys=False).apply(_build_lags)

# Pick POI with most rows
counts = df["poi_id"].value_counts()
poi_sel = POI_ID if POI_ID is not None else (counts.index[0] if not counts.empty else None)
hist = df[df["poi_id"]==poi_sel].sort_values("date").copy()
hist = hist.dropna(subset=["lag_1","lag_7","lag_14","lag_28","roll_mean_7","roll_mean_28","roll_std_7","roll_std_28"])

# Limit to last WINDOW_DAYS
if len(hist):
    cutoff = hist["date"].max() - pd.Timedelta(days=WINDOW_DAYS)
    hist = hist[hist["date"]>=cutoff].copy()

tr, te = _time_split(hist, test_size=0.2)

engineered = set(["lag_1","lag_7","lag_14","lag_28","roll_mean_7","roll_mean_28","roll_std_7","roll_std_28"]) | set(CAL_KEEP)
top10 = _top10_original_numeric(tr, exclude=engineered)

feat_cols = list(dict.fromkeys(CAL_KEEP + list(engineered - set(CAL_KEEP)) + top10))

gb = HistGradientBoostingRegressor(
    learning_rate=0.06,
    max_iter=400,
    max_leaf_nodes=31,
    early_stopping=True,
    validation_fraction=0.12,
    random_state=42
)
gb.fit(tr[feat_cols], tr["count"])

pred_tr = gb.predict(tr[feat_cols])
pred_te = gb.predict(te[feat_cols])

r2_tr = r2_score(tr["count"], pred_tr)
mae_tr = mean_absolute_error(tr["count"], pred_tr)
r2_te = r2_score(te["count"], pred_te)
mae_te = mean_absolute_error(te["count"], pred_te)

print(f"POI: {poi_sel}")
print(f"Top-10 base numeric: {top10}")
print(f"Features used: {len(feat_cols)}")
print("")
print(f"Train R^2: {r2_tr:.4f} | Train MAE: {mae_tr:.2f}")
print(f" Test R^2: {r2_te:.4f} |  Test MAE: {mae_te:.2f}")
```

```python
# ===== Fast variant (complete, end-to-end) =====
# - Limits training window to last WINDOW_DAYS for the selected POI
# - Same calendar + lag/rolling features as your snippet
# - HistGradientBoostingRegressor
# - Recursive future prediction (keeps features identical & ordered)
import pandas as pd, numpy as np, warnings
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import r2_score, mean_absolute_error

# -------- CONFIG --------
CSV_PATH = r"C:\Users\israt\HD lab\My Project\Dataset\Airsage Dataset\NebraskaLakes_Cleaned_round.csv"
POI_ID = None          # set to an int to force a POI; or None -> auto-pick with most rows
FORECAST_DAYS = 14
WINDOW_DAYS = 800      # training window length (days from last date)

# Keep EXACT same calendar features
CAL_KEEP = [
    "month","day","dow","week","is_weekend","season",
    "sin_doy","cos_doy","sin_dow","cos_dow"
]

# -------- helpers: calendar + engineering --------
def _season_from_month(m):
    if m in (12,1,2): return 1
    if m in (3,4,5):  return 2
    if m in (6,7,8):  return 3
    return 4

def _make_calendar(d):
    d = d.copy()
    d["month"] = d["date"].dt.month.astype(np.int16)
    d["day"] = d["date"].dt.day.astype(np.int16)
    d["dow"] = d["date"].dt.weekday.astype(np.int16)
    d["week"] = d["date"].dt.isocalendar().week.astype(np.int16)
    d["is_weekend"] = d["dow"].isin([5,6]).astype(np.int8)
    d["season"] = d["month"].apply(_season_from_month).astype(np.int8)
    doy = d["date"].dt.dayofyear.astype(int)
    d["sin_doy"] = np.sin(2*np.pi*doy/365).astype(np.float32)
    d["cos_doy"] = np.cos(2*np.pi*doy/365).astype(np.float32)
    d["sin_dow"] = np.sin(2*np.pi*(d["dow"]+1)/7).astype(np.float32)
    d["cos_dow"] = np.cos(2*np.pi*(d["dow"]+1)/7).astype(np.float32)
    return d

def _build_lags(g):
    g = g.sort_values("date").copy()
    for k in [1,7,14,28]:
        g[f"lag_{k}"] = g["count"].shift(k)
    g["roll_mean_7"]  = g["count"].rolling(7,  min_periods=1).mean().shift(1)
    g["roll_mean_28"] = g["count"].rolling(28, min_periods=1).mean().shift(1)
    g["roll_std_7"]   = g["count"].rolling(7,  min_periods=1).std(ddof=0).shift(1)
    g["roll_std_28"]  = g["count"].rolling(28, min_periods=1).std(ddof=0).shift(1)
    return g

def _time_split(h, test_size=0.2):
    n = len(h)
    n_te = max(1, int(round(n*test_size)))
    n_tr = n - n_te
    return h.iloc[:n_tr].copy(), h.iloc[n_tr:].copy()

def _top10_original_numeric(tr, target="count", exclude=()):
    num_cols = tr.select_dtypes(include=[np.number]).columns.tolist()
    ex = set(exclude) | {target, "poi_id"}
    base = [c for c in num_cols if c not in ex]
    if not base:
        return []
    corr = tr[base].corrwith(tr[target]).abs().sort_values(ascending=False)
    return corr.head(10).index.tolist()

def _poi_name_from_df(df_all, poi_id):
    if "poi" in df_all.columns:
        s = df_all.loc[df_all["poi_id"] == poi_id, "poi"].dropna()
        if not s.empty:
            return str(s.iloc[-1])
    return f"POI_{poi_id}"

# -------- load & prep --------
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    df = pd.read_csv(CSV_PATH, low_memory=False)

# required columns
if "date" not in df.columns or "count" not in df.columns:
    raise ValueError("CSV must include 'date' and 'count' columns (and ideally 'poi' or 'poi_id').")

# parse/clean
df["date"] = pd.to_datetime(df["date"].astype(str), format="%Y%m%d", errors="coerce")
df = df.dropna(subset=["date"])
df["count"] = pd.to_numeric(df["count"], errors="coerce")
df = df.dropna(subset=["count"])

# ensure poi_id
if "poi_id" not in df.columns:
    if "poi" in df.columns:
        df["poi_id"] = pd.factorize(df["poi"].astype(str))[0].astype(int)
    else:
        df["poi_id"] = 0

df = df.sort_values(["poi_id","date"]).reset_index(drop=True)
df = _make_calendar(df)
df = df.groupby("poi_id", group_keys=False).apply(_build_lags)

# -------- select POI + filter training window --------
counts = df["poi_id"].value_counts()
poi_sel = POI_ID if POI_ID is not None else (counts.index[0] if not counts.empty else None)
if poi_sel is None:
    raise ValueError("No POIs found in the data.")

hist = df[df["poi_id"] == poi_sel].sort_values("date").copy()

# need complete engineered features before training
need_cols = ["lag_1","lag_7","lag_14","lag_28","roll_mean_7","roll_mean_28","roll_std_7","roll_std_28"]
hist = hist.dropna(subset=need_cols)

if hist.empty:
    raise ValueError(f"POI {poi_sel} has no rows after lag/rolling filtering.")

# limit to last WINDOW_DAYS
cutoff = hist["date"].max() - pd.Timedelta(days=WINDOW_DAYS)
hist = hist[hist["date"] >= cutoff].copy()

if len(hist) < 50:
    warnings.warn(f"Only {len(hist)} rows for POI {poi_sel} in window; results may be unstable.")

# -------- train/test split + features --------
tr, te = _time_split(hist, test_size=0.2)

engineered = set(need_cols) | set(CAL_KEEP)
top10 = _top10_original_numeric(tr, exclude=engineered)

# exact feature order: calendar -> engineered lags/rollings -> top10 extras
feat_cols = list(dict.fromkeys(
    CAL_KEEP + list(engineered - set(CAL_KEEP)) + top10
))

# -------- model --------
gb = HistGradientBoostingRegressor(
    learning_rate=0.06,
    max_iter=400,
    max_leaf_nodes=31,
    early_stopping=True,
    validation_fraction=0.12,
    random_state=42
)
gb.fit(tr[feat_cols], tr["count"])

# -------- metrics --------
pred_tr = gb.predict(tr[feat_cols])
pred_te = gb.predict(te[feat_cols])
r2_tr = r2_score(tr["count"], pred_tr)
mae_tr = mean_absolute_error(tr["count"], pred_tr)
r2_te = r2_score(te["count"], pred_te)
mae_te = mean_absolute_error(te["count"], pred_te)

print(f"POI (id): {poi_sel} | name: {_poi_name_from_df(df, poi_sel)}")
print(f"Top-10 base numeric: {top10}")
print(f"Features used: {len(feat_cols)} -> {feat_cols}\n")
print(f"Train R^2: {r2_tr:.4f} | Train MAE: {mae_tr:.2f}")
print(f" Test R^2: {r2_te:.4f} |  Test MAE: {mae_te:.2f}")

# -------- future prediction (recursive, SAME features/order) --------
def _future_calendar(start_date, days):
    d = pd.DataFrame({"date": pd.date_range(start_date, periods=days, freq="D")})
    return _make_calendar(d)

def predict_future_same_feats(hist_df, model, feat_cols, poi_id, days=14):
    """
    hist_df: per-POI history AFTER engineering and dropna on needed lags/rollings
    model:   trained regressor
    feat_cols: EXACT feature order used for training
    poi_id:  int POI id
    days:    horizon
    Returns: DataFrame with poi_id, date, date_str, weekday, predicted
    """
    if hist_df.empty:
        raise ValueError("No history for this POI after lag/rolling filtering.")

    h = hist_df.sort_values("date").copy()
    last_date = h["date"].max()
    fut_cal = _future_calendar(last_date + pd.Timedelta(days=1), days)

    # init with last 28 actual counts
    recent = h["count"].tail(28).astype(float).tolist()
    rows = []

    for i in range(days):
        row = fut_cal.iloc[[i]].copy()

        # lags
        def L(k): return recent[-k] if len(recent) >= k else np.nan
        row["lag_1"]  = L(1)
        row["lag_7"]  = L(7)
        row["lag_14"] = L(14)
        row["lag_28"] = L(28)

        arr = np.asarray(recent, dtype=float)
        if len(arr) > 0:
            row["roll_mean_7"]  = (arr[-7:].mean()  if len(arr) >= 7  else arr.mean())
            row["roll_mean_28"] = (arr[-28:].mean() if len(arr) >= 28 else arr.mean())
            row["roll_std_7"]   = (arr[-7:].std(ddof=0)  if len(arr) >= 7  else arr.std(ddof=0))
            row["roll_std_28"]  = (arr[-28:].std(ddof=0) if len(arr) >= 28 else arr.std(ddof=0))
        else:
            row["roll_mean_7"] = row["roll_mean_28"] = 0.0
            row["roll_std_7"]  = row["roll_std_28"]  = 0.0

        # fallback for first days
        for c in ["lag_7","lag_14","lag_28","roll_mean_28","roll_std_28"]:
            if pd.isna(row[c]).iloc[0]:
                row[c] = row["lag_1"] if not pd.isna(row["lag_1"]).iloc[0] else 0.0

        # SAME feature order
        Xf = row.reindex(columns=feat_cols, fill_value=0)
        yhat = float(model.predict(Xf)[0])
        recent.append(yhat)

        rows.append({
            "poi_id": int(poi_id),
            "date": row["date"].iloc[0],
            "predicted": yhat
        })

    out = pd.DataFrame(rows).sort_values("date")
    out["date_str"] = out["date"].dt.strftime("%Y-%m-%d")
    out["weekday"] = out["date"].dt.day_name()
    return out

# -------- run forecast + format exactly as requested --------
future_pred = predict_future_same_feats(hist, gb, feat_cols, poi_sel, days=FORECAST_DAYS)

poi_name = _poi_name_from_df(df, poi_sel)
out = future_pred.copy()
out["POI NAME"]   = poi_name
out["Prediction"] = out["predicted"]                # use .round(0).astype(int) if you want integers
out["DAY"]        = out["date"].dt.day_name()
out["Month"]      = out["date"].dt.month_name()
out["Date"]       = out["date"].dt.strftime("%Y-%m-%d")

final_view = out[["POI NAME", "Prediction", "DAY", "Month", "Date"]]
print("\n--- Future predictions ---")
print(final_view.to_string(index=False))
```

```python
def predict_future_same_feats(hist_df, model, feat_cols, poi_id, days=14):
    """
    Builds future rows using SAME feature order as training.
    For any extra columns (not calendar/lag/rolling), forward-fill from the
    last known values in hist_df instead of using zeros.
    """
    if hist_df.empty:
        raise ValueError("No history for this POI after lag/rolling filtering.")

    # Define engineered sets we can compute at forecast time
    cal_set = {
        "month","day","dow","week","is_weekend","season",
        "sin_doy","cos_doy","sin_dow","cos_dow"
    }
    lagset = {"lag_1","lag_7","lag_14","lag_28"}
    rollset = {"roll_mean_7","roll_mean_28","roll_std_7","roll_std_28"}
    engineered_all = cal_set | lagset | rollset

    # Last known per-feature values from history (for extra features)
    h = hist_df.sort_values("date").copy()
    last_vals = h.tail(1).reindex(columns=feat_cols)  # aligns to feat_cols
    # Fallback medians if some columns are NaN at the last row
    medians = h[feat_cols].median(numeric_only=True)
    # Map of extra (non-engineered) columns that must be forward-filled
    extra_cols = [c for c in feat_cols if c not in engineered_all]

    last_date = h["date"].max()
    fut_cal = _future_calendar(last_date + pd.Timedelta(days=1), days)

    # Seed recent counts with as much actual history as possible (up to 56)
    seed_depth = min(56, len(h))
    recent = h["count"].tail(seed_depth).astype(float).tolist()

    rows = []
    for i in range(days):
        row = fut_cal.iloc[[i]].copy()  # calendar features already set

        # ---- lags from recent buffer ----
        def L(k): return recent[-k] if len(recent) >= k else np.nan
        row["lag_1"]  = L(1)
        row["lag_7"]  = L(7)
        row["lag_14"] = L(14)
        row["lag_28"] = L(28)

        # ---- rollings from recent buffer (no zero fill) ----
        arr = np.asarray(recent, dtype=float)
        if len(arr) >= 2:
            row["roll_mean_7"]  = (arr[-7:].mean()  if len(arr) >= 7  else arr.mean())
            row["roll_mean_28"] = (arr[-28:].mean() if len(arr) >= 28 else arr.mean())
            row["roll_std_7"]   = (arr[-7:].std(ddof=0)  if len(arr) >= 7  else arr.std(ddof=0))
            row["roll_std_28"]  = (arr[-28:].std(ddof=0) if len(arr) >= 28 else arr.std(ddof=0))
        else:
            # use the most recent value as mean, and 0 std if only one value
            v = arr[-1] if len(arr) == 1 else np.nan
            row["roll_mean_7"]  = v
            row["roll_mean_28"] = v
            row["roll_std_7"]   = 0.0 if len(arr) == 1 else np.nan
            row["roll_std_28"]  = 0.0 if len(arr) == 1 else np.nan

        # ---- graceful fallback for early steps (no hard zeros) ----
        if pd.isna(row["lag_1"]).iloc[0]:
            # if absolutely no seed, fall back to last known count from history
            row["lag_1"] = float(h["count"].iloc[-1])
        for c in ["lag_7","lag_14","lag_28"]:
            if pd.isna(row[c]).iloc[0]:
                row[c] = row["lag_1"]

        for c in ["roll_mean_7","roll_mean_28"]:
            if pd.isna(row[c]).iloc[0]:
                row[c] = row["lag_1"]
        for c in ["roll_std_7","roll_std_28"]:
            if pd.isna(row[c]).iloc[0]:
                row[c] = 0.0

        # ---- bring in EXTRA feature columns by forward-filling from history ----
        if extra_cols:
            # take last known values for extras; if NaN, use column median
            extra_df = last_vals[extra_cols].copy()
            for c in extra_cols:
                v = extra_df.iloc[0][c]
                if pd.isna(v):
                    v = medians.get(c, 0.0)
                row[c] = v

        # ---- enforce SAME feature order ----
        Xf = row.reindex(columns=feat_cols, fill_value=0)

        yhat = float(model.predict(Xf)[0])
        # counts should be non-negative
        if np.isnan(yhat):
            yhat = float(row["lag_1"])
        yhat = max(0.0, yhat)

        recent.append(yhat)
        rows.append({
            "poi_id": int(poi_id),
            "date": row["date"].iloc[0],
            "predicted": yhat
        })

    out = pd.DataFrame(rows).sort_values("date")
    out["date_str"] = out["date"].dt.strftime("%Y-%m-%d")
    out["weekday"] = out["date"].dt.day_name()
    return out
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error
from xgboost import XGBRegressor

# ======================
# Step 0: Load & prepare
# ======================
df = pd.read_csv(
    r"C:\Users\israt\HD lab\My Project\Dataset\Airsage Dataset\NebraskaLakes_Cleaned_round.csv",
    low_memory=False
)

# Clean types
df['date'] = pd.to_datetime(df['date'], format='%Y%m%d', errors='coerce')
df['count'] = pd.to_numeric(df['count'], errors='coerce')
df['duration'] = pd.to_numeric(df['duration'], errors='coerce')
df = df.dropna(subset=['date', 'count', 'duration']).copy()

# ======================
# Step A: Calendar features (ordered) + Season
# ======================
df['month'] = df['date'].dt.month.astype(np.int16)
df['dow'] = df['date'].dt.dayofweek.astype(np.int16)  # 0=Mon .. 6=Sun
df['is_weekend'] = (df['dow'] >= 5).astype(np.int8)

# Season: 0=Winter(Dec–Feb), 1=Spring(Mar–May), 2=Summer(Jun–Aug), 3=Fall(Sep–Nov)
def _season_from_month(m: int) -> int:
    if m in (12, 1, 2):  return 0
    if m in (3, 4, 5):   return 1
    if m in (6, 7, 8):   return 2
    return 3
df['season'] = df['month'].map(_season_from_month).astype(np.int8)

TIME_FEATS = ['month', 'dow', 'is_weekend', 'season']  # maintain order

# ======================
# Step B: Anomaly features (same as before)
# ======================
X_ad = StandardScaler().fit_transform(df[['count', 'duration']])
iso = IsolationForest(contamination=0.10, random_state=42)
pred = iso.fit_predict(X_ad)
score = -iso.decision_function(X_ad)

df['anomaly_label'] = (pred == -1).astype(np.int8)   # 1 = anomaly
df['anomaly_score'] = score.astype(np.float32)

# ======================
# Step C: Build feature matrix (ALL NUMERIC FEATURES, with TIME_FEATS first)
# ======================
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

# Target
TARGET = 'count'

# Put TIME_FEATS first (if present), then the rest (excluding TARGET and duplicates)
ordered_prefix = [c for c in TIME_FEATS if c in numeric_cols]
rest = [c for c in numeric_cols if c not in set(ordered_prefix + [TARGET])]
feature_cols = ordered_prefix + rest

X_all = df[feature_cols].fillna(0)
y = df[TARGET].values.astype(np.float32)

# Keep side columns for segmented metrics on test split
side = df[['dow', 'is_weekend', 'season']].reset_index(drop=True)

# Train/val split (random for now; switch to time-based split if needed)
X_train, X_test, y_train, y_test, side_tr, side_te = train_test_split(
    X_all, y, side, test_size=0.20, random_state=42
)

# ======================
# Step D: XGBoost training (same hyperparams)
# ======================
xgb = XGBRegressor(
    n_estimators=600,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    random_state=42,
    n_jobs=-1
)
xgb.fit(X_train, y_train)

# ======================
# Step E: Evaluation — overall + weekday/weekend + season breakdown
# ======================
y_tr_pred = xgb.predict(X_train)
y_te_pred = xgb.predict(X_test)

print("=== XGBoost with ordered time-features (month, dow, is_weekend, season) ===")
print("Train R²:", round(r2_score(y_train, y_tr_pred), 4))
print("Test  R²:", round(r2_score(y_test, y_te_pred), 4))
print("Test  MAE:", round(mean_absolute_error(y_test, y_te_pred), 3))

```

```python
import pandas as pd
import numpy as np

# ======================
# Load & basic prep
# ======================
PATH = r"C:\Users\israt\HD lab\My Project\Dataset\Airsage Dataset\NebraskaLakes_Cleaned_round.csv"

df = pd.read_csv(PATH, low_memory=False)

# clean types
df['date'] = pd.to_datetime(df['date'].astype(str), format='%Y%m%d', errors='coerce')
df['count'] = pd.to_numeric(df['count'], errors='coerce')
df = df.dropna(subset=['date', 'count']).copy()

# required cols
if 'poi_id' not in df.columns:
    raise ValueError("Dataset must contain a 'poi_id' column.")

# time features
df['month'] = df['date'].dt.month.astype('int16')
df['dow'] = df['date'].dt.dayofweek.astype('int16')  # 0=Mon..6=Sun
df['is_weekend'] = (df['dow'] >= 5).astype('int8')

def _season_from_month(m: int) -> int:
    # 0=DJF, 1=MAM, 2=JJA, 3=SON
    if m in (12, 1, 2):  return 0
    if m in (3, 4, 5):   return 1
    if m in (6, 7, 8):   return 2
    return 3

# guarantee season exists
df['season'] = df['month'].map(_season_from_month).astype('int8')

# ======================
# Build simple pattern lookups
# ======================
# Core pattern table: POI × (month, dow) → median count
poi_m_dow = df.groupby(['poi_id', 'month', 'dow'])['count'].median().rename('med_poi_m_dow')
poi_m     = df.groupby(['poi_id', 'month'])['count'].median().rename('med_poi_m')
poi_dow   = df.groupby(['poi_id', 'dow'])['count'].median().rename('med_poi_dow')
poi_all   = df.groupby(['poi_id'])['count'].median().rename('med_poi_all')

# Global fallbacks
g_m_dow = df.groupby(['month', 'dow'])['count'].median().rename('med_g_m_dow')
g_m     = df.groupby(['month'])['count'].median().rename('med_g_m')
g_dow   = df.groupby(['dow'])['count'].median().rename('med_g_dow')
g_all   = float(df['count'].median())

# Season scalers (per-POI, relative to its overall median) — keeps broad seasonal shape
poi_season = (df.groupby(['poi_id', 'season'])['count'].mean()
                .rename('mean_poi_season')).to_frame()
poi_overall = df.groupby('poi_id')['count'].mean().rename('mean_poi_all')
poi_season = poi_season.join(poi_overall, on='poi_id')
poi_season['season_ratio'] = (poi_season['mean_poi_season'] / poi_season['mean_poi_all']).clip(0.5, 1.5)
poi_season = poi_season['season_ratio']  # keep only ratio

# Global season ratios (fallback)
g_season = (df.groupby('season')['count'].mean() / df['count'].mean()).clip(0.5, 1.5)

# Weekend uplift (learned from data), with global fallback and sane clipping
wk_tbl = df.groupby(['poi_id','is_weekend'])['count'].median().unstack()  # cols: {0,1}
poi_weekend_uplift = ((wk_tbl.get(1) / wk_tbl.get(0)).clip(0.8, 2.0)).rename('poi_weekend_uplift')

w_med = df.loc[df['is_weekend']==1, 'count'].median()
wd_med = df.loc[df['is_weekend']==0, 'count'].median()
g_weekend_uplift = (w_med / wd_med) if (pd.notna(w_med) and pd.notna(wd_med) and wd_med != 0) else 1.2
g_weekend_uplift = float(np.clip(g_weekend_uplift, 0.8, 2.0))

# ======================
# Forecast function (simple & robust) + weekend spike
# ======================
def forecast_visitors(df, poi_id, start_date, days=7, blend_season: float = 1.0,
                      weekend_amp: float = 1.0,      # 0=no extra, 1=full learned uplift
                      min_weekend_boost: float = 1.10 # at least +10% on weekends
                     ):
    """
    Forecast future 'count' for a POI for N days starting at start_date (YYYYMMDD or YYYY-MM-DD).
    Pattern = POI median by (month, dow) with fallbacks × season ratio × WEEKEND SPIKE.

    weekend_amp: scales how strongly to apply the weekend uplift (learned from data).
                 Example: uplift=1.3, weekend_amp=1.0 -> 1.30x; weekend_amp=0.5 -> 1.15x.
    min_weekend_boost: floor for weekend multiplier (e.g., 1.10 = at least +10%).
    """
    # Prepare future calendar
    start = pd.to_datetime(str(start_date), format='%Y%m%d', errors='coerce')
    if pd.isna(start):
        start = pd.to_datetime(str(start_date), errors='coerce')
    if pd.isna(start):
        raise ValueError("start_date must be in format YYYYMMDD or YYYY-MM-DD")

    future_dates = pd.date_range(start=start, periods=int(days), freq='D')
    fut = pd.DataFrame({'date': future_dates})
    fut['month'] = fut['date'].dt.month.astype('int16')
    fut['dow'] = fut['date'].dt.dayofweek.astype('int16')
    fut['season'] = fut['month'].map(_season_from_month).astype('int8')
    fut['is_weekend'] = (fut['dow'] >= 5).astype('int8')

    # Helper to fetch baseline from cascaded medians
    def _baseline_for_row(row):
        m = int(row.month); d = int(row.dow)
        # Try POI-level detailed pattern first
        v = poi_m_dow.get((poi_id, m, d), np.nan)
        if not pd.isna(v): return float(v)
        # Cascade fallbacks
        for series, key in [(poi_m, (poi_id, m)), (poi_dow, (poi_id, d))]:
            v = series.get(key, np.nan)
            if not pd.isna(v): return float(v)
        v = poi_all.get(poi_id, np.nan)
        if not pd.isna(v): return float(v)
        # Global fallbacks
        v = g_m_dow.get((m, d), np.nan)
        if not pd.isna(v): return float(v)
        v = g_m.get(m, np.nan)
        if not pd.isna(v): return float(v)
        v = g_dow.get(d, np.nan)
        if not pd.isna(v): return float(v)
        return float(g_all)

    # Season multiplier (per POI → else global)
    def _season_mult(sea):
        v = poi_season.get((poi_id, int(sea)), np.nan)
        if pd.isna(v):
            v = float(g_season.get(int(sea), 1.0))
        # blend toward 1.0 for simplicity control
        return float((1 - blend_season) * 1.0 + blend_season * v)

    # Weekend multiplier for this POI (fallback to global)
    _uplift = float(poi_weekend_uplift.get(poi_id, np.nan))
    if not np.isfinite(_uplift):
        _uplift = g_weekend_uplift

    def _wk_mult(is_weekend: int) -> float:
        if is_weekend:
            # 1 + amp*(uplift-1)  -> amplify or soften the learned uplift
            mult = 1.0 + weekend_amp * (_uplift - 1.0)
            return float(max(mult, min_weekend_boost))
        return 1.0

    # Compute predictions
    base = fut.apply(_baseline_for_row, axis=1).to_numpy(dtype=float)
    seas = fut['season'].map(_season_mult).to_numpy(dtype=float)
    wmul = fut['is_weekend'].map(_wk_mult).to_numpy(dtype=float)
    pred = (base * seas * wmul).astype('float32')

    out = fut.copy()
    out['poi_id'] = poi_id
    out['Prediction'] = pred
    out['date_str'] = out['date'].dt.strftime('%Y%m%d')
    out['weekday'] = out['date'].dt.day_name()
    out['Month'] = out['date'].dt.strftime('%b')
    return out[['poi_id','date_str','weekday','Month','month','dow','season','is_weekend','Prediction']]

# ======================
# Example usage
# ======================
if __name__ == "__main__":
    # Example: forecast 14 days for a POI starting from 2025-01-07
    POI = 4731
    START = "20250107"
    DAYS = 14

    fut = forecast_visitors(
        df, POI, START, days=DAYS,
        blend_season=1.0,          # 0..1 season strength
        weekend_amp=1.0,           # 0=no extra, 1=full learned uplift
        min_weekend_boost=1.20     # ensure ≥ +20% on weekends
    )
    # Optional rounding to whole people:
    fut['Prediction'] = fut['Prediction'].round(0)

    print(fut.to_string(index=False))
```

```python
import pandas as pd
import numpy as np

# ======================
# Load & basic prep
# ======================
PATH = r"C:\Users\israt\HD lab\My Project\Dataset\Airsage Dataset\NebraskaLakes_Cleaned_round.csv"

df = pd.read_csv(PATH, low_memory=False)

# clean types
df['date']  = pd.to_datetime(df['date'].astype(str), format='%Y%m%d', errors='coerce')
df['count'] = pd.to_numeric(df['count'], errors='coerce')
df = df.dropna(subset=['date', 'count']).copy()

if 'poi_id' not in df.columns:
    raise ValueError("Dataset must contain a 'poi_id' column.")

# time features
df['month']      = df['date'].dt.month.astype('int16')
df['dow']        = df['date'].dt.dayofweek.astype('int16')  # 0=Mon..6=Sun
df['is_weekend'] = (df['dow'] >= 5).astype('int8')

def _season_from_month(m: int) -> int:
    # 0=DJF, 1=MAM, 2=JJA, 3=SON
    if m in (12, 1, 2):  return 0
    if m in (3, 4, 5):   return 1
    if m in (6, 7, 8):   return 2
    return 3
df['season'] = df['month'].map(_season_from_month).astype('int8')

# ======================
# Pattern lookups (medians) + season ratios
# ======================
# POI medians
poi_m_dow = df.groupby(['poi_id', 'month', 'dow'])['count'].median()
poi_m     = df.groupby(['poi_id', 'month'])['count'].median()
poi_dow   = df.groupby(['poi_id', 'dow'])['count'].median()
poi_all   = df.groupby(['poi_id'])['count'].median()

# Global fallbacks
g_m_dow = df.groupby(['month', 'dow'])['count'].median()
g_m     = df.groupby(['month'])['count'].median()
g_dow   = df.groupby(['dow'])['count'].median()
g_all   = float(df['count'].median())

# Per-POI season ratios (relative to POI mean)
poi_season = (df.groupby(['poi_id', 'season'])['count'].mean()
                .rename('mean_poi_season')).to_frame()
poi_overall = df.groupby('poi_id')['count'].mean().rename('mean_poi_all')
poi_season = poi_season.join(poi_overall, on='poi_id')
poi_season['season_ratio'] = (poi_season['mean_poi_season'] / poi_season['mean_poi_all']).clip(0.5, 1.5)
poi_season = poi_season['season_ratio']  # MultiIndex (poi_id, season) -> ratio

# Global season ratios (fallback)
g_season = (df.groupby('season')['count'].mean() / df['count'].mean()).clip(0.5, 1.5)

# Weekend uplift learned from data (weekend / weekday), with global fallback
wk_tbl = df.groupby(['poi_id','is_weekend'])['count'].median().unstack()  # columns {0,1}
poi_weekend_uplift = ((wk_tbl.get(1) / wk_tbl.get(0))).clip(0.8, 2.0)

w_med  = df.loc[df['is_weekend']==1, 'count'].median()
wd_med = df.loc[df['is_weekend']==0, 'count'].median()
g_weekend_uplift = (w_med / wd_med) if (pd.notna(w_med) and pd.notna(wd_med) and wd_med!=0) else 1.2
g_weekend_uplift = float(np.clip(g_weekend_uplift, 0.8, 2.0))

# ======================
# Forecast (with weekend spike + locked Fri/Sat/Sun pattern)
# ======================
def forecast_visitors(df, poi_id, start_date, days=7,
                      blend_season: float = 1.0,     # 0..1 strength of season multiplier
                      weekend_amp: float = 1.0,      # 0=no extra, 1=full learned uplift
                      min_weekend_boost: float = 1.10, # ≥ +10% on weekends
                      weekend_floor_rel: float = 1.25, # weekends ≥ 25% above weekday ref
                      weekend_floor_mode: str = "median", # "median" | "max" | "friday"
                      # Locked pattern knobs:
                      friday_min_rel: float = 1.10,   # Friday ≥ +10% vs weekday ref
                      friday_max_rel: float = 1.22,   # Friday ≤ +22% vs weekday ref
                      sat_floor_rel:  float = 1.30,   # Saturday ≥ +30% vs weekday ref
                      sun_floor_rel:  float = 1.25,   # Sunday   ≥ +25% vs weekday ref
                      weekday_cap_rel: float = None   # cap Mon–Thu (and Mon–Fri) ≤ rel * weekday ref (None = off)
                     ):
    """
    Prediction = baseline(POI medians w/ fallbacks) × season_ratio × weekend_multiplier
    + floors/caps to lock Friday moderate and Sat/Sun strong spikes.
    """
    # ---- future calendar ----
    start = pd.to_datetime(str(start_date), format='%Y%m%d', errors='coerce')
    if pd.isna(start):
        start = pd.to_datetime(str(start_date), errors='coerce')
    if pd.isna(start):
        raise ValueError("start_date must be YYYYMMDD or YYYY-MM-DD")

    fut = pd.DataFrame({'date': pd.date_range(start=start, periods=int(days), freq='D')})
    fut['month']      = fut['date'].dt.month.astype('int16')
    fut['dow']        = fut['date'].dt.dayofweek.astype('int16')
    fut['season']     = fut['month'].map(_season_from_month).astype('int8')
    fut['is_weekend'] = (fut['dow'] >= 5).astype('int8')

    # ---- baseline from cascaded medians ----
    def _baseline_for_row(row):
        m = int(row.month); d = int(row.dow)
        v = poi_m_dow.get((poi_id, m, d), np.nan)
        if pd.notna(v): return float(v)
        for ser, key in [(poi_m,(poi_id,m)), (poi_dow,(poi_id,d))]:
            v = ser.get(key, np.nan)
            if pd.notna(v): return float(v)
        v = poi_all.get(poi_id, np.nan)
        if pd.notna(v): return float(v)
        # globals
        v = g_m_dow.get((m,d), np.nan)
        if pd.notna(v): return float(v)
        v = g_m.get(m, np.nan)
        if pd.notna(v): return float(v)
        v = g_dow.get(d, np.nan)
        if pd.notna(v): return float(v)
        return float(g_all)

    # ---- season multiplier ----
    def _season_mult(sea):
        v = poi_season.get((poi_id, int(sea)), np.nan)
        if pd.isna(v):
            v = float(g_season.get(int(sea), 1.0))
        return float((1 - blend_season) * 1.0 + blend_season * v)

    # ---- weekend multiplier (learned uplift, scaled, with minimum) ----
    _uplift = float(poi_weekend_uplift.get(poi_id, np.nan)) if poi_id in poi_weekend_uplift.index else np.nan
    if not np.isfinite(_uplift):
        _uplift = g_weekend_uplift

    def _wk_mult(is_weekend: int) -> float:
        if is_weekend:
            mult = 1.0 + weekend_amp * (_uplift - 1.0)
            return float(max(mult, min_weekend_boost))
        return 1.0

    # ---- compute prediction parts ----
    base = fut.apply(_baseline_for_row, axis=1).to_numpy(float)
    seas = fut['season'].map(_season_mult).to_numpy(float)
    wmul = fut['is_weekend'].map(_wk_mult).to_numpy(float)
    prelim = base * seas * wmul  # before floors/caps

    # ---------- Lock the pattern: Fri moderate, Sat/Sun strong ----------
    dow = fut['dow'].to_numpy()
    wk_mask  = (dow >= 5)
    fri_mask = (dow == 4)
    sat_mask = (dow == 5)
    sun_mask = (dow == 6)
    wd_mask  = (dow < 5)

    # Weekday reference in this window
    if weekend_floor_mode == "friday" and np.any(dow == 4):
        weekday_ref = float(prelim[dow == 4].mean())
    elif wd_mask.any():
        weekday_ref = float(np.max(prelim[wd_mask]) if weekend_floor_mode == "max"
                            else np.median(prelim[wd_mask]))
    else:
        weekday_ref = float(np.median(prelim))

    # 1) Apply general weekend floor (both Sat & Sun)
    prelim[wk_mask] = np.maximum(prelim[wk_mask], weekend_floor_rel * weekday_ref)

    # 2) Friday moderate band (min and max)
    if fri_mask.any():
        prelim[fri_mask] = np.clip(prelim[fri_mask],
                                   friday_min_rel * weekday_ref,
                                   friday_max_rel * weekday_ref)

    # 3) Saturday & Sunday stronger floors
    if sat_mask.any():
        prelim[sat_mask] = np.maximum(prelim[sat_mask], sat_floor_rel * weekday_ref)
    if sun_mask.any():
        prelim[sun_mask] = np.maximum(prelim[sun_mask], sun_floor_rel * weekday_ref)

    # 4) Optional cap for all weekdays (keeps Mon–Thu from jumping over Friday)
    if weekday_cap_rel is not None and wd_mask.any():
        prelim[wd_mask] = np.minimum(prelim[wd_mask], weekday_cap_rel * weekday_ref)

    # ---- finalize ----
    out = fut.copy()
    out['poi_id']     = poi_id
    out['Prediction'] = prelim.astype('float32')
    out['date_str']   = out['date'].dt.strftime('%Y%m%d')
    out['weekday']    = out['date'].dt.day_name()
    out['Month']      = out['date'].dt.strftime('%b')
    return out[['poi_id','date_str','weekday','Month','month','dow','season','is_weekend','Prediction']]

# ======================
# Example usage
# ======================
if __name__ == "__main__":
    POI   = 4731
    START = "20250107"
    DAYS  = 14

    fut = forecast_visitors(
        df, POI, START, days=DAYS,
        blend_season=1.0,          # 0..1 strength of seasonality
        weekend_amp=1.0,           # learned uplift scaling for weekends
        min_weekend_boost=1.20,    # weekends at least +20% over weekday ref
        weekend_floor_rel=1.25,    # weekends ≥ 25% above weekday ref
        weekend_floor_mode="median",
        friday_min_rel=1.10,       # Friday 10% above weekday ref
        friday_max_rel=1.22,       # Friday ≤ 22% above weekday ref
        sat_floor_rel=1.30,        # Saturday ≥ 30% above weekday ref
        sun_floor_rel=1.25,        # Sunday   ≥ 25% above weekday ref
        weekday_cap_rel=1.22       # cap weekdays to Friday’s upper band
    )
    # Optional rounding to whole people:
    fut['Prediction'] = fut['Prediction'].round(0)
    print(fut.to_string(index=False))
```

```python
import pandas as pd
import numpy as np

# ======================
# CONFIG
# ======================
PATH = r"C:\Users\israt\HD lab\My Project\Dataset\Airsage Dataset\NebraskaLakes_Cleaned_round.csv"
ID_COL, DATE_COL, TARGET = "poi_id", "date", "count"

# Range construction
P_LOW, P_HIGH = 0.10, 0.90     # robust day range (P10–P90)
MAX_RATIO     = 2.0            # cap overly wide bands: hi <= lo * MAX_RATIO
MIN_WIDTH     = 1.0            # minimum band width (absolute)

# Sampling
DEFAULT_SEED  = 42             # None -> different every run
SAMPLER       = "uniform"      # "uniform" or "triangular"

# Pattern controls (per ISO week, post-sample)
FRI_MIN_REL   = 1.10           # Friday ≥ 10% above weekday reference
FRI_MAX_REL   = 1.22           # Friday ≤ 22% above weekday reference
SAT_FLOOR_REL = 1.30           # Saturday ≥ 30% above weekday reference
SUN_FLOOR_REL = 1.25           # Sunday   ≥ 25% above weekday reference
SUN_CAP_REL   = 1.00           # Sunday ≤ 1.00 × Saturday (set 1.05 to allow slight Sun>Sat)
SAT_BIAS      = 1.00           # multiply Saturday prediction (e.g., 1.02) to favor Sat peak

# Weekday smoothing (kills Monday spike)
WKDAY_CAP_REL = 1.22           # cap Mon–Thu (and Mon–Fri below Friday band top)
MONDAY_CAP_REL= 1.05           # Monday ≤ 5% above weekday reference

# ======================
# LOAD & PREP
# ======================
df = pd.read_csv(PATH, low_memory=False)
df[DATE_COL] = pd.to_datetime(df[DATE_COL].astype(str), format="%Y%m%d", errors="coerce")
df[TARGET]   = pd.to_numeric(df[TARGET], errors="coerce")
df = df.dropna(subset=[DATE_COL, TARGET]).copy()
if ID_COL not in df.columns:
    raise ValueError("Need 'poi_id' column")

df["month"] = df[DATE_COL].dt.month.astype("int8")
df["dow"]   = df[DATE_COL].dt.dayofweek.astype("int8")        # 0=Mon..6=Sun
df["dom"]   = df[DATE_COL].dt.day.astype("int8")
df["wom"]   = ((df["dom"] - 1) // 7 + 1).clip(1, 5).astype("int8")  # 1..5

# ======================
# RANGES (P10–P90) with first-week preference
# ======================
def q(g, p): return g.quantile(p)

pmwd = df.groupby([ID_COL, "month", "wom", "dow"])[TARGET].agg(lo=lambda s: q(s, P_LOW), hi=lambda s: q(s, P_HIGH))
pm1d = df[df["wom"]==1].groupby([ID_COL, "month", "dow"])[TARGET].agg(lo=lambda s: q(s, P_LOW), hi=lambda s: q(s, P_HIGH))
pmd  = df.groupby([ID_COL, "month", "dow"])[TARGET].agg(lo=lambda s: q(s, P_LOW), hi=lambda s: q(s, P_HIGH))
pdw  = df.groupby([ID_COL, "dow"])[TARGET].agg(lo=lambda s: q(s, P_LOW), hi=lambda s: q(s, P_HIGH))
gmd  = df.groupby(["month", "dow"])[TARGET].agg(lo=lambda s: q(s, P_LOW), hi=lambda s: q(s, P_HIGH))
g_lo, g_hi = float(df[TARGET].quantile(P_LOW)), float(df[TARGET].quantile(P_HIGH))

def _safe_band(lo, hi):
    lo = float(lo) if np.isfinite(lo) else g_lo
    hi = float(hi) if np.isfinite(hi) else g_hi
    if hi < lo: lo, hi = hi, lo
    if lo > 0 and hi > lo * MAX_RATIO:
        hi = lo * MAX_RATIO
    if hi - lo < MIN_WIDTH:
        mid = 0.5 * (lo + hi)
        lo, hi = max(0.0, mid - MIN_WIDTH/2), max(0.0, mid + MIN_WIDTH/2)
    return lo, hi

def _lookup_range(poi_id, month, wom, dow):
    for table, key in [(pmwd,(poi_id,month,wom,dow)),
                       (pm1d,(poi_id,month,dow)),
                       (pmd, (poi_id,month,dow)),
                       (pdw, (poi_id,dow)),
                       (gmd, (month,dow))]:
        try:
            r = table.loc[key]; return _safe_band(r["lo"], r["hi"])
        except KeyError:
            pass
    return _safe_band(g_lo, g_hi)

# ======================
# FORECAST (range sample + weekly pattern enforcement)
# ======================
def forecast_range_sample(poi_id, start_date, days=14, seed=DEFAULT_SEED, sampler=SAMPLER):
    rng = np.random.default_rng(seed)
    start = pd.to_datetime(str(start_date), format="%Y%m%d", errors="coerce")
    if pd.isna(start): start = pd.to_datetime(str(start_date), errors="coerce")
    if pd.isna(start): raise ValueError("start_date must be YYYYMMDD or YYYY-MM-DD")

    future = pd.date_range(start, periods=int(days), freq="D")
    rows = []
    for d in future:
        m = int(d.month)
        w = int(((d.day-1)//7)+1); w = 1 if w<1 else 5 if w>5 else w
        wd= int(d.weekday())
        lo, hi = _lookup_range(poi_id, m, w, wd)
        if sampler == "triangular":
            y = rng.triangular(lo, 0.5*(lo+hi), hi)
        else:
            y = rng.uniform(lo, hi)
        rows.append({"poi_id":poi_id, "date":d, "date_str":d.strftime("%Y%m%d"),
                     "weekday":d.day_name(), "Month":d.strftime("%b"),
                     "month":m, "wom":w, "dow":wd, "is_weekend":int(wd>=5),
                     "lo":lo, "hi":hi, "Prediction":float(y)})
    out = pd.DataFrame(rows)

    # ---------- WEEKLY PATTERN ENFORCEMENT (kills Monday spike, shapes Fri/Sat/Sun) ----------
    # ISO week ids
    iso = out["date"].dt.isocalendar()
    out["iso_year"] = iso["year"].astype(int)
    out["iso_week"] = iso["week"].astype(int)

    for (yy, ww), idx in out.groupby(["iso_year","iso_week"]).groups.items():
        g = out.loc[list(idx)]

        # Weekday reference from Tue–Thu if present, else all weekdays
        tuth = g["dow"].isin([1,2,3])
        wk   = g["dow"] < 5
        if tuth.any():
            wref = float(g.loc[tuth, "Prediction"].median())
        elif wk.any():
            wref = float(g.loc[wk, "Prediction"].median())
        else:
            wref = float(g["Prediction"].median())

        # Friday band
        fri = g["dow"] == 4
        if fri.any():
            fmin, fmax = FRI_MIN_REL*wref, FRI_MAX_REL*wref
            out.loc[fri.index[fri], "Prediction"] = np.clip(g.loc[fri, "Prediction"], fmin, fmax)
            # reflect band in hi/lo
            out.loc[fri.index[fri], "hi"] = np.minimum(out.loc[fri.index[fri], "hi"], fmax)
            out.loc[fri.index[fri], "lo"] = np.minimum(out.loc[fri.index[fri], "lo"], out.loc[fri.index[fri], "hi"])

        # Weekday caps (prevents Monday spike)
        wk_mask = g["dow"] < 5
        if wk_mask.any():
            cap_all = WKDAY_CAP_REL * wref
            out.loc[wk_mask.index[wk_mask], "Prediction"] = np.minimum(g.loc[wk_mask, "Prediction"], cap_all)
            # specific Monday tighter cap
            mon = g["dow"] == 0
            if mon.any():
                mon_cap = min(cap_all, MONDAY_CAP_REL * wref)
                out.loc[mon.index[mon], "Prediction"] = np.minimum(out.loc[mon.index[mon], "Prediction"], mon_cap)
            # reflect caps in hi
            out.loc[wk_mask.index[wk_mask], "hi"] = np.minimum(out.loc[wk_mask.index[wk_mask], "hi"], cap_all)
            if mon.any():
                out.loc[mon.index[mon], "hi"] = np.minimum(out.loc[mon.index[mon], "hi"], mon_cap)
            # keep lo <= hi
            sel = wk_mask | mon
            hi_vals = out.loc[sel.index[sel], "hi"]
            lo_vals = out.loc[sel.index[sel], "lo"]
            out.loc[sel.index[sel], "lo"] = np.minimum(lo_vals, hi_vals)

        # Saturday floor (+ optional bias)
        sat = g["dow"] == 5
        if sat.any():
            sfloor = SAT_FLOOR_REL * wref
            out.loc[sat.index[sat], "Prediction"] = np.maximum(out.loc[sat.index[sat], "Prediction"], sfloor)
            out.loc[sat.index[sat], "Prediction"] *= SAT_BIAS
            out.loc[sat.index[sat], "lo"] = np.maximum(out.loc[sat.index[sat], "lo"], sfloor)
            out.loc[sat.index[sat], "hi"] = np.maximum(out.loc[sat.index[sat], "hi"], out.loc[sat.index[sat], "lo"] + MIN_WIDTH)

        # Sunday floor + cap to Saturday
        sun = g["dow"] == 6
        if sun.any() and sat.any():
            # cap
            sat_val = float(out.loc[sat.index[sat], "Prediction"].iloc[0])
            sun_cap = sat_val * SUN_CAP_REL
            out.loc[sun.index[sun], "Prediction"] = np.minimum(out.loc[sun.index[sun], "Prediction"], sun_cap)
            out.loc[sun.index[sun], "hi"] = np.minimum(out.loc[sun.index[sun], "hi"], sun_cap)
        if sun.any():
            sfloor = SUN_FLOOR_REL * wref
            out.loc[sun.index[sun], "Prediction"] = np.maximum(out.loc[sun.index[sun], "Prediction"], sfloor)
            out.loc[sun.index[sun], "lo"] = np.maximum(out.loc[sun.index[sun], "lo"], sfloor)
            out.loc[sun.index[sun], "hi"] = np.maximum(out.loc[sun.index[sun], "hi"], out.loc[sun.index[sun], "lo"] + MIN_WIDTH)

        # Ensure predictions lie within (lo, hi); if not, re-sample uniformly within band
        sel = out.loc[list(idx)]
        mask = (sel["Prediction"] < sel["lo"]) | (sel["Prediction"] > sel["hi"])
        if mask.any():
            i = sel.index[mask]
            lo = out.loc[i, "lo"].to_numpy()
            hi = out.loc[i, "hi"].to_numpy()
            resampled = lo + (hi - lo) * np.random.default_rng(seed=DEFAULT_SEED).random(len(i))
            out.loc[i, "Prediction"] = resampled

    # tidy
    out = out.drop(columns=["iso_year","iso_week"], errors="ignore")
    out["lo"]  = out["lo"].round(2)
    out["hi"]  = out["hi"].round(2)
    out["Prediction"] = out["Prediction"].astype(float)
    return out[["poi_id","date_str","weekday","Month","month","wom","dow","is_weekend","lo","hi","Prediction"]]

# ======================
# EXAMPLE
# ======================
if __name__ == "__main__":
    POI, START, DAYS = 4731, "20250707", 14
    fut = forecast_range_sample(POI, START, days=DAYS, seed=42, sampler="uniform")
    fut["Prediction"] = fut["Prediction"].round(0)  # whole people
    print(fut.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
from decimal import Decimal, InvalidOperation

# ====== Config ======
PATH = r"C:\Users\israt\HD lab\My Project\Dataset\Airsage Dataset\NebraskaLakes_Cleaned_round.csv"
POI = "4731"               # your POI
BASE = "20220708"          # anchor (use its month/day for every year)
DAYS = 14                  # window length (days)
PRED_YEAR = 2025           # predicted year to compare against 2022/23/24

# ====== POI key normalizer (robust) ======
def norm_poi_key(x):
    if x is None:
        return None
    s = str(x).strip().replace(",", "")
    if s == "":
        return None
    try:
        d = Decimal(s)
    except InvalidOperation:
        return s
    if d == d.to_integral():
        return str(int(d))
    return format(d.normalize(), "f")

# ====== Load & prep ======
df = pd.read_csv(PATH, low_memory=False)

# required cols
for col in ["poi_id", "date", "count"]:
    if col not in df.columns:
        raise ValueError(f"Missing column '{col}' in dataset")

# types
df["date"] = pd.to_datetime(df["date"], format="%Y%m%d", errors="coerce")
df["count"] = pd.to_numeric(df["count"], errors="coerce")
df = df.dropna(subset=["date", "count"]).copy()

# normalize POI keys
df["poi_key"] = df["poi_id"].apply(norm_poi_key)
poi_key = norm_poi_key(POI)

# one row per (poi_key, date) using median (robust to duplicates)
day_med = (df.groupby(["poi_key", "date"], as_index=False)["count"]
             .median()
             .rename(columns={"count": "count_med"}))

# add helpers for a simple predictor
df["month"] = df["date"].dt.month.astype("int8")
df["dow"]   = df["date"].dt.dayofweek.astype("int8")  # 0=Mon..6=Sun

# ====== helpers ======
def window_for_year(poi_key: str, year: int, base_yyyymmdd: str, days: int) -> pd.DataFrame:
    """Return a DAYS-length window for 'year' aligned to BASE's month/day."""
    base_dt = pd.to_datetime(str(year) + base_yyyymmdd[4:], format="%Y%m%d")
    cal = pd.DataFrame({"date": pd.date_range(base_dt, periods=days, freq="D")})
    w = cal.merge(
        day_med[day_med["poi_key"] == poi_key][["date", "count_med"]],
        on="date", how="left"
    )
    w["date_str"] = w["date"].dt.strftime("%Y%m%d")
    w["weekday"]  = w["date"].dt.day_name()
    return w[["date", "date_str", "weekday", "count_med"]].rename(columns={"count_med": f"y{year}"})

def simple_predict_for_year(poi_key: str, year: int, base_yyyymmdd: str, days: int) -> pd.DataFrame:
    """
    Simple, non-flat predictor:
      baseline = median(count | POI, month, dow) with global fallback
      weekend bump = (weekend_median / weekday_median), clipped
      small ±7.5% jitter so values vary
    """
    base_dt = pd.to_datetime(str(year) + base_yyyymmdd[4:], format="%Y%m%d")
    cal = pd.DataFrame({"date": pd.date_range(base_dt, periods=days, freq="D")})
    cal["month"] = cal["date"].dt.month.astype("int8")
    cal["dow"]   = cal["date"].dt.dayofweek.astype("int8")
    cal["is_weekend"] = (cal["dow"] >= 5).astype("int8")

    poi_hist = df[df["poi_key"] == poi_key]
    poi_m_d  = poi_hist.groupby(["month","dow"])["count"].median()
    g_m_d    = df.groupby(["month","dow"])["count"].median()

    def baseline(m, d):
        v = poi_m_d.get((m, d), np.nan)
        if not pd.isna(v): return float(v)
        v = g_m_d.get((m, d), np.nan)
        if not pd.isna(v): return float(v)
        return float(df["count"].median())

    # weekend uplift (POI -> global -> default)
    if not poi_hist.empty:
        wk = poi_hist.copy()
        wk["is_weekend"] = (wk["dow"] >= 5).astype("int8")
        wd_med = wk[wk["is_weekend"]==0]["count"].median()
        we_med = wk[wk["is_weekend"]==1]["count"].median()
        poi_uplift = (we_med / wd_med) if pd.notna(we_med) and pd.notna(wd_med) and wd_med != 0 else np.nan
    else:
        poi_uplift = np.nan
    if not np.isfinite(poi_uplift):
        wd_med_g = df[df["dow"]<5]["count"].median()
        we_med_g = df[df["dow"]>=5]["count"].median()
        poi_uplift = (we_med_g / wd_med_g) if pd.notna(we_med_g) and pd.notna(wd_med_g) and wd_med_g != 0 else 1.2
    poi_uplift = float(np.clip(poi_uplift, 1.05, 1.8))

    rng = np.random.default_rng(42)  # reproducible jitter
    preds = []
    for _, r in cal.iterrows():
        m, d = int(r["month"]), int(r["dow"])
        b = baseline(m, d)
        wmul = poi_uplift if d >= 5 else 1.0
        y = b * wmul
        y *= (1.0 + rng.uniform(-0.075, 0.075))  # ±7.5% jitter
        preds.append(max(0.0, y))

    out = cal.copy()
    out["y_pred"] = preds
    out["date_str"] = out["date"].dt.strftime("%Y%m%d")
    out["weekday"]  = out["date"].dt.day_name()
    return out[["date", "date_str", "weekday", "y_pred"]]

# ====== Build aligned windows ======
w22 = window_for_year(poi_key, 2022, BASE, DAYS).rename(columns={
    "weekday": "weekday_2022", "date": "date_2022", "date_str": "date_str_2022"
})
w23 = window_for_year(poi_key, 2023, BASE, DAYS).rename(columns={
    "weekday": "weekday_2023", "date": "date_2023", "date_str": "date_str_2023"
})
w24 = window_for_year(poi_key, 2024, BASE, DAYS).rename(columns={
    "weekday": "weekday_2024", "date": "date_2024", "date_str": "date_str_2024"
})
wp  = simple_predict_for_year(poi_key, PRED_YEAR, BASE, DAYS).rename(columns={
    "weekday": "weekday_2025", "date": "date_2025", "date_str": "date_str_2025", "y_pred": "y2025_pred"
})

# add position key (0..DAYS-1) and merge by position
for _df in (w22, w23, w24, wp):
    _df["k"] = np.arange(len(_df), dtype=int)

cmp = (
    wp[["k","date_str_2025","weekday_2025","y2025_pred"]]
      .merge(w22[["k","date_str_2022","weekday_2022","y2022"]], on="k", how="left")
      .merge(w23[["k","date_str_2023","weekday_2023","y2023"]], on="k", how="left")
      .merge(w24[["k","date_str_2024","weekday_2024","y2024"]], on="k", how="left")
      .drop(columns="k")
)

# Optional rounding to whole people
for c in ["y2022","y2023","y2024","y2025_pred"]:
    if c in cmp.columns:
        cmp[c] = cmp[c].round(0)

print(f"\n=== POI {POI} | Aligned {DAYS}-day window: 2022 vs 2023 vs 2024 vs Pred({PRED_YEAR}) ===")
print(cmp.to_string(index=False))
```
